### Smooth data using LOWESS (Locally Weighted Regression), plot smoothed and original data, plot just smoothed data
Adjust *frac* value in the LOWESS funcion to change smoothing by adjusting the *smooth_level* variable below.  A higher value results in more smoothing.  Typical values for *frac* are from 0.02 to 0.1. Increase slightly if you still see noise.

```python
from statsmodels.nonparametric.smoothers_lowess import lowess

smooth_level=0.1 # <-- Adjust this value to change smoothing, 0.02–0.1 typical

smoothed = lowess(pu_0_0_df['Stress (MPa)'],
                  pu_0_0_df['Strain (%)'],
                  frac=smooth_level)

smoothed_data = pd.DataFrame({
    'Strain (%)': smoothed[:, 0],
    'Stress (MPa)': smoothed[:, 1]
})

# Plot the original data
sns.lineplot(x='Strain (%)', y='Stress (MPa)', data=pu_0_0_df, label='Original Data')

#Plot the smoothed data
sns.lineplot(x='Strain (%)',
             y='Stress (MPa)',
             data=smoothed_data,
             label=f'Smoothed Data (frac={smooth_level})')

plt.title('Stress vs Strain for PU_0_0: Original vs Smoothed Data')
plt.legend() # Show legend to distinguish the lines
plt.show()
sns.lineplot(x='Strain (%)', y='Stress (MPa)', data=smoothed_data)
plt.show()
```

### Polynomial fit (degree 5) of the data.

```python
# Polynomial fit through the origin
def poly_fit_through_origin(x, y, degree):
    # To force the polynomial through the origin (0,0), we fit a polynomial
    # of the form y = a_n * x^n + a_{n-1} * x^{n-1} + ... + a_1 * x
    # This is equivalent to fitting y/x = a_n * x^{n-1} + ... + a_1
    # However, dividing by x is problematic when x is close to zero.
    # A more robust approach is to use linear algebra to solve for the coefficients
    # with the constraint that the constant term is zero.

    # Construct the Vandermonde matrix for the polynomial fit without the constant term
    # The columns will be x^degree, x^(degree-1), ..., x^1
    X = np.vander(x, degree + 1)[:, :-1] # Exclude the x^0 (constant) term

    # Use least squares to find the coefficients
    # We want to solve X @ coeffs = y for coeffs
    coeffs, residuals, rank, s = np.linalg.lstsq(X, y, rcond=None)

    # The coefficients are for x^degree, x^(degree-1), ..., x^1
    # We need to create a polynomial function. np.poly1d expects coefficients
    # in descending powers, including the constant term. Since our constant is 0,
    # we can append a 0 to the coefficients.
    poly_coeffs = np.append(coeffs, 0)

    # Create the polynomial function
    poly = np.poly1d(poly_coeffs)

    return poly

# Apply polynomial fit to all dataframes
poly_degree = 5  # Use the same degree as before

dataframes = {
    'PU 0.0': pu_0_0_df,
    'PU 0.1': pu_0_1_df,
    'PU 0.2': pu_0_2_df,
    'PU 0.3': pu_0_3_df
}

smoothed_poly_dataframes = {}

for name, df in dataframes.items():
    poly_fit = poly_fit_through_origin(df['Strain (%)'], df['Stress (MPa)'], degree=poly_degree)
    # Generate smoothed data points using the polynomial fit
    strain_smoothed = np.insert(df['Strain (%)'].values, 0, 0)
    stress_smoothed = poly_fit(strain_smoothed)

    smoothed_poly_dataframes[name] = pd.DataFrame({
        'Strain (%)': strain_smoothed,
        'Stress (MPa)': stress_smoothed,
        'Sample': name # Add a column to identify the sample
    })

# Concatenate all smoothed dataframes for plotting
all_smoothed_poly_data = pd.concat(smoothed_poly_dataframes.values())

# Plot original and smoothed data for all samples
plt.figure(figsize=(9, 6))

# Plot original data
for name, df in dataframes.items():
    sns.lineplot(x='Strain (%)', y='Stress (MPa)', data=df, label=f'{name} (Original)', linestyle='--', alpha=0.5) # Add alpha for transparency

# Plot smoothed data
sns.lineplot(x='Strain (%)', y='Stress (MPa)', hue='Sample', data=all_smoothed_poly_data)


plt.title('Stress vs Strain: Original vs Polynomial Smoothed Data (Through Origin)')
plt.xlabel('Strain (%)')
plt.ylabel('Stress (MPa)')
plt.legend()
plt.grid(True)
plt.show()

# Optional: Plot only the smoothed data for better clarity
plt.figure(figsize=(9, 6))
sns.lineplot(x='Strain (%)', y='Stress (MPa)', hue='Sample', data=all_smoothed_poly_data)

plt.title('Stress vs Strain: Polynomial Smoothed Data (Through Origin)')
plt.xlabel('Strain (%)')
plt.ylabel('Stress (MPa)')
plt.grid(True)
plt.show()
```

# Try fitting with a spline.  

This code snippet uses `scipy.interpolate.UnivariateSpline` to apply spline smoothing to the 'Stress (MPa)' data of `df1`.

*   `UnivariateSpline(x, y, s=0.1)` creates a cubic spline fit to the data points. The `s` parameter is the smoothing factor, which controls the trade-off between fidelity to the data and smoothness of the curve. A smaller `s` means the spline will pass closer to the original points (less smoothing), while a larger `s` will result in a smoother curve that might deviate more from the original points.
*   `np.linspace` is used to create a denser set of x-values, and `spline(x_smooth)` evaluates the spline at these new points to generate a smooth curve.
*   The plot shows both the original data points and the resulting spline-smoothed curve for comparison.*italicized text*

```python
import numpy as np
from scipy.interpolate import UnivariateSpline
import matplotlib.pyplot as plt

# Select a DataFrame to demonstrate spline smoothing
df_to_smooth = pu_0_0_df.copy()

# Get Strain and Stress data
x = df_to_smooth['Strain (%)']
y = df_to_smooth['Stress (MPa)']

# Create a spline object. 's' is the smoothing factor.
# A larger 's' value means more smoothing. Adjust as needed.
# You might need to experiment with 's' to get the desired smoothness.
spline = UnivariateSpline(x, y, s=0.1) # 's' can be adjusted

# Generate new x values for a smoother curve, if desired
x_smooth = np.linspace(x.min(), x.max(), 500)
y_smooth = spline(x_smooth)

# Plot original and smoothed data
plt.figure(figsize=(10, 6))
plt.plot(x, y, 'o', label='Original pu_0_0_df Data', markersize=4, alpha=0.6)
plt.plot(x_smooth, y_smooth, label='pu_0_0_df Spline Smoothed Data', color='red')
plt.xlabel('Strain (%)')
plt.ylabel('Stress (MPa)')
plt.title('Spline Smoothing Example for pu_0_0_df')
plt.legend()
plt.grid(True)
plt.show()
```

Fit all four curves with Spline.  Force spline through the origin (0,0)

```python
import numpy as np
from scipy.interpolate import UnivariateSpline
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

plt.figure(figsize=(12, 7))

dataframes = {'df0': pu_0_0_df, 'df1': pu_0_1_df, 'df2': pu_0_2_df, 'df3': pu_0_3_df}

smoothing_factors = {
    'df0': 0.7,
    'df1': 0.7,
    'df2': 0.7,
    'df3': 0.7
}

smoothed_data_for_seaborn = pd.DataFrame()
palette = sns.color_palette('deep', n_colors=len(dataframes)) # Get the palette once

# Collect handles and labels for custom legend
all_handles = []
all_labels = []

for i, (name, df) in enumerate(dataframes.items()):
    x_np = df['Strain (%)'].values
    y_np = df['Stress (MPa)'].values

    # Prepare data for spline fitting, forcing through origin
    y_for_spline = np.copy(y_np)
    weights = np.ones_like(y_np)

    if x_np[0] == 0.0:
        y_for_spline[0] = 0.0
        weights[0] = 1e10

    # Apply spline smoothing
    spline = UnivariateSpline(x_np, y_for_spline, w=weights, s=smoothing_factors[name])
    x_smooth = np.linspace(x_np.min(), x_np.max(), 500)
    y_smooth = spline(x_smooth)

    # Prepare smoothed data for seaborn
    temp_df = pd.DataFrame({
        'Strain (%)': x_smooth,
        'Stress (MPa)': y_smooth,
        'DataFrame': name
    })
    smoothed_data_for_seaborn = pd.concat([smoothed_data_for_seaborn, temp_df])

    # Plot original data points with matching color and transparency
    line, = plt.plot(x_np, y_np, 'o', color=palette[i], alpha=0.5, markersize=4)
    all_handles.append(line)
    all_labels.append(f'Original {name}')

# Plot smoothed data using seaborn
sns_plot = sns.lineplot(
    data=smoothed_data_for_seaborn,
    x='Strain (%)',
    y='Stress (MPa)',
    hue='DataFrame',
    palette=palette # Explicitly pass the palette to ensure color consistency
)

# Get handles and labels for the seaborn lines and combine them with original data
sns_handles, sns_labels = sns_plot.get_legend_handles_labels()

# The first label from sns_labels is typically the hue column name itself ('DataFrame'), skip it.
# The subsequent labels are the actual hue values ('df1', 'df2', etc.)
for j in range(0, len(sns_labels)):
    all_handles.append(sns_handles[j])
    all_labels.append(f'{sns_labels[j]} (Smoothed)')

# Remove the default seaborn legend
sns_plot.legend_.remove()

plt.xlabel('Strain (%)')
plt.ylabel('Stress (MPa)')
plt.title('Spline Smoothed Stress vs. Strain with Original Data (Forced through Origin)')
plt.grid(True)

# Create a single combined legend outside the plot
plt.legend(all_handles, all_labels, title='Data Type', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout() # Adjust layout to prevent legend overlap
plt.show()
```

```python
import numpy as np
from scipy.interpolate import UnivariateSpline
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Set up matplotlib for publication quality aesthetics
plt.rcParams.update({
    'font.size': 14, # Increased font size
    'axes.labelsize': 16, # Increased label size
    'axes.titlesize': 18, # Increased title size
    'xtick.labelsize': 14, # Increased tick label size
    'ytick.labelsize': 14, # Increased tick label size
    'figure.titlesize': 20 # Increased figure title size
})

plt.figure(figsize=(10, 8))

dataframes = {'df0': pu_0_0_df, 'df1': pu_0_1_df, 'df2': pu_0_2_df, 'df3': pu_0_3_df}

smoothing_factors = {
    'df0': 0.7,
    'df1': 0.7,
    'df2': 0.7,
    'df3': 0.7
}

smoothed_data_for_seaborn = pd.DataFrame()
palette = sns.color_palette('deep', n_colors=len(dataframes)) # Get the palette once

label_map = {
    'df0': 'PU 0.0',
    'df1': 'PU 0.1',
    'df2': 'PU 0.2',
    'df3': 'PU 0.3'
}

# Store smoothed curve data for later direct annotation
smoothed_curves_for_annotation = {}

for i, (name, df) in enumerate(dataframes.items()):
    x_np = df['Strain (%)'].values
    y_np = df['Stress (MPa)'].values

    # Prepare data for spline fitting, forcing through origin
    y_for_spline = np.copy(y_np)
    weights = np.ones_like(y_np)

    if x_np[0] == 0.0:
        y_for_spline[0] = 0.0
        weights[0] = 1e10

    # Apply spline smoothing
    spline = UnivariateSpline(x_np, y_for_spline, w=weights, s=smoothing_factors[name])
    x_smooth = np.linspace(x_np.min(), x_np.max(), 500)
    y_smooth = spline(x_smooth)

    # Prepare smoothed data for seaborn (using original df names for hue color mapping)
    temp_df = pd.DataFrame({
        'Strain (%)': x_smooth,
        'Stress (MPa)': y_smooth,
        'DataFrame': name
    })
    smoothed_data_for_seaborn = pd.concat([smoothed_data_for_seaborn, temp_df])

    # Store the smoothed curve data along with its assigned color for later annotation
    smoothed_curves_for_annotation[name] = {'x': x_smooth, 'y': y_smooth, 'color': palette[i]}

# Plot smoothed data using seaborn
ax = sns.lineplot(
    data=smoothed_data_for_seaborn,
    x='Strain (%)',
    y='Stress (MPa)',
    hue='DataFrame',
    palette=palette,
    linewidth=2.5 # Increased line thickness
)

# Remove the default seaborn legend
ax.legend_.remove()

# Remove gridlines
ax.grid(False)

plt.xlabel('Strain (%)')
plt.ylabel('Stress (MPa)')
# plt.title('Spline Smoothed Stress vs. Strain') # Removed title

# Set X and Y axis ranges
plt.xlim(0, 600)
plt.ylim(0, 3)

# plt.gca().set_aspect('equal', adjustable='box') # Removed line to make aspect ratio square

# Add custom labels directly to the lines
for name, data in smoothed_curves_for_annotation.items():
    # Find a good position for the label (e.g., near the end of the line)
    x_pos = data['x'][-1]
    y_pos = data['y'][-1]

    # Small manual adjustments for better label positioning if lines are close
    if name == 'df0':
        y_pos += 0.15
        x_pos += 20 # Move df0 label 20 units to the right
    elif name == 'df1':
        y_pos += 0.10
        x_pos += 20 # Move df1 label 20 units to the right
    elif name == 'df2':
        y_pos += 0.08  # Keep vertical adjustment
        x_pos += 20    # Keep df2 label 20 units to the right
    elif name == 'df3':
        y_pos += 0.12
        x_pos += 20 # Move df3 label 20 units to the right

    plt.text(x_pos, y_pos, label_map[name], color=data['color'],
             ha='right', va='center', fontsize=13) # Increased fontsize for labels

plt.tight_layout() # Adjust layout to prevent elements from overlapping
plt.savefig('/content/GraftedHybridLatexPaper/SVG_Outputs/Fig2a_stress_strain.svg') # Save the plot as an SVG file
plt.show()
```
